### 내장함수 range

```python
a = range(10)
print("range(10) => value : {} / type : {}".format(a, type(a)))

for i in a:
    print(i, end=",")
```

```python
b = range(2,10)
print("range(2,10) => value : {} ".format(b))

for i in b:
    print(i, end=",")
```

```python
c = range(8,1,-2)
print(f"range(8,1,-2) => value : {c}")

for i in c:
    print(i, end=",")
```
